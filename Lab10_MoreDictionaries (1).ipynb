```python
def max_wind(data):
    obs = data['obs']
    fastest_wind = -1
    for key in obs:
        if obs[key]['w_spd'] > fastest_wind:
            fastest_wind = obs[key]['w_spd']
            press = key
    return fastest_wind, press

print(max_wind(stations["KGFK"]))
print(max_wind(stations["KMOT"]))




```

```python
def lapse_rate(observations):
    lapse_rate = 0

    for level in observations:
        if level == 'sfc':
            t_sfc = observations[level]['temp']
            z_sfc = observations[level]['height']
        if level == '900':
            t_900 = observations[level]['temp']
            z_900 = observations[level]['height']

    lapse = (t_900 - t_sfc) / (z_900 - z_sfc)
    
    return lapse


print(lapse_rate(stations["KFSD"]["obs"]))
print(lapse_rate(stations["KRST"]["obs"]))
```

4) Now let's figure out which station has the most stable air. We can do this by looking at the lapse rate of each station. We know that the lapse rate is greater than 9.8 C per km the air is unstable and if it is less than 9.8 C per km the air is stable. Finish the function below to calcualte the lapse rate of each station, then return the most stable and most unstable stations and thier lapse rates. The function stability will take the stations dictionary as an argument. Which station is most stable and which station is least stable? What are their lapse rates?

```python
def stability(stations):
    max_stab = ("", -1000)
    min_stab = ("", 1000)

    for key in stations:
        obs = stations[key]['obs']
        stab = obs['sfc']['temp'] - obs['500']['temp']

        if stab > max_stab[1]:
            max_stab = (key, stab)
        if stab < min_stab[1]:
            min_stab = (key, stab)
    
    most_station = stations[max_stab[0]]
    least_station = stations[min_stab[0]]
    
def lapse_rate(station):
    obs = station['obs']
    t_sfc = obs['sfc']['temp']
    t_500 = obs['500']['temp']
    z_sfc = obs['sfc']['height']
    z_500 = obs['500']['height']
    return (t_500 - t_sfc) / (z_500 - z_sfc)
    
print(f"Most stable station: {most_station['name']} (Stability = {max_stab[1]}°C)")
print(f"Lapse rate: {lapse_rate(most_station):.2f}°C/km\n")

print(f"Least stable station: {least_station['name']} (Stability = {min_stab[1]}°C)")
print(f"Lapse rate: {lapse_rate(least_station):.2f}°C/km")
```
